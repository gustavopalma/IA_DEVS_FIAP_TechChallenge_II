Como segunda etapa da exploração dos dados, buscamos identificar as principais correlações entre as variáveis. Para isso, utilizamos o método corr() da biblioteca Pandas que, por padrão, utiliza a correlação de Pearson. Essa medida estatística indica a direção e a intensidade da associação linear entre duas variáveis. Seu coeficiente varia entre -1 e 1: valores próximos de 1 indicam uma forte associação linear positiva, enquanto valores próximos de -1 indicam uma forte associação linear negativa. Valores próximos de zero indicam pouca ou nenhuma associação linear. É importante ressaltar que a existência de correlação não implica uma relação de causalidade entre as variáveis.

```python
#Análise Exploratória de dados
def greaterCorrelations(df, limit=20):
    corr = df.corr(numeric_only=True)

    
    mask = np.triu(
        np.ones(corr.shape),
        k=1
    ).astype(bool)

    corrs = corr.where(mask).stack()

    
    indices = corrs.abs().sort_values(
        ascending=False
    ).index

    resultado = corrs.loc[indices].head(limit)

    return resultado

greaterCorrelations(data)
```

Mesmo que a análise anterior permita identificar relações fortes entre quaisquer variáveis do conjunto de dados, nosso objetivo no projeto é investigar características associadas à presença de SOP. Dessa forma, é necessário analisar especificamente as correlações entre cada variável e a variável-alvo armazenada na coluna "PCOS (Y/N)".

```python
def correlationsWithTarget(df, target):
    corr = df.corr(numeric_only=True)

    result = (
        corr[target]
        .drop(target)
        .sort_values(
            key=abs,
            ascending=False
        )
    )

    return result

correlationsWithTarget(data,"PCOS (Y/N)")
```

Dando sequência a nossa análise, podemos analisar alguns sintomas clínicos e a relação deles com o diagnóstico ou não de SOP. Lembrando que a presença destes sintomas não significam que o paciente possui SOP, mas sim que geralmente o paciente que apresenta o diagnóstico possui essas características.

```python
features = [
    "Skin darkening (Y/N)",
    "hair growth(Y/N)",
    "Weight gain(Y/N)",
    "Pimples(Y/N)"
]

labels = {
    "Skin darkening (Y/N)": "Escurecimento da pele",
    "hair growth(Y/N)": "Crescimento de pelos",
    "Weight gain(Y/N)": "Ganho de peso",
    "Pimples(Y/N)": "Acne"
}

resultado = []

for feature in features:

    sem_sop = (
        data[data["PCOS (Y/N)"] == 0][feature].mean()
        * 100
    )

    com_sop = (
        data[data["PCOS (Y/N)"] == 1][feature].mean()
        * 100
    )

    resultado.append({
        "Característica": labels[feature],
        "Sem SOP": sem_sop,
        "Com SOP": com_sop
    })

clinical = pd.DataFrame(resultado)

ax = clinical.set_index("Característica").plot(
    kind="bar",
    figsize=(10, 6)
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=3
    )

plt.title("Manifestações clínicas em pacientes com e sem SOP")
plt.xlabel("")
plt.ylabel("Pacientes (%)")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

O gráfico apresenta diferenças expressivas na frequência das manifestações clínicas entre os dois grupos. Entre as pacientes diagnosticadas com SOP, 62,5% apresentam escurecimento da pele, 57,4% crescimento de pelos, 68,8% ganho de peso e 69,9% acne. Em todos os casos, os percentuais são superiores aos encontrados no grupo sem SOP. Entretanto, essa análise considera cada característica isoladamente e não controla outros fatores que também podem influenciar essas manifestações, como idade, alimentação ou outras condições clínicas. Portanto, os resultados demonstram associação com o diagnóstico de SOP, e não uma relação de causalidade.

## Otimização de hiperparâmetros por Algoritmos Genéticos

Nesta etapa, será implementado um algoritmo genético para otimizar
os hiperparâmetros dos modelos de classificação utilizados na Fase 1.

Cada indivíduo representa uma configuração de hiperparâmetros,
e uma população corresponde a um conjunto de configurações candidatas.

### Representação dos indivíduos e operadores genéticos

O algoritmo foi implementado no módulo `src/algo_genetico.py`. Cada indivíduo da população representa uma configuração possível da Árvore de Decisão, composta pelos seguintes genes:

- `max_depth`: profundidade máxima da árvore;
- `min_samples_split`: número mínimo de amostras necessário para dividir um nó;
- `min_samples_leaf`: número mínimo de amostras em uma folha;
- `criterion`: critério utilizado para avaliar as divisões (`gini` ou `entropy`).

A população inicial é gerada aleatoriamente dentro dos intervalos definidos para cada parâmetro. Em seguida, o algoritmo aplica **seleção por torneio**, **cruzamento de genes**, **mutação** e **elitismo**, preservando o melhor indivíduo identificado ao longo da execução.

A função de aptidão (*fitness*) corresponde à média do **F1-score** obtido em uma validação cruzada estratificada de três partições, aplicada somente ao conjunto de treinamento. Essa estratégia permite comparar as configurações candidatas sem utilizar o conjunto de teste durante a otimização.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

melhor_individuo = melhores[notas.index(max(notas))]

criterio = "gini" if melhor_individuo.criterion == 0 else "entropy"

modelo_otimizado = DecisionTreeClassifier(
    max_depth=melhor_individuo.max_depth,
    min_samples_split=melhor_individuo.min_samples_split,
    min_samples_leaf=melhor_individuo.min_samples_leaf,
    criterion=criterio,
    random_state=42
)

modelo_otimizado.fit(X_train, y_train)

y_pred_otimizado = modelo_otimizado.predict(X_test)

modelo_original = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

modelo_original.fit(X_train, y_train)

y_pred_original = modelo_original.predict(X_test)

import pandas as pd

def avaliar(y_real, y_pred):
    return {
        "Accuracy": accuracy_score(y_real, y_pred),
        "Precision": precision_score(y_real, y_pred),
        "Recall": recall_score(y_real, y_pred),
        "F1-score": f1_score(y_real, y_pred)
    }

comparacao = pd.DataFrame({
    "Original": avaliar(y_test, y_pred_original),
    "Otimizado": avaliar(y_test, y_pred_otimizado)
})

comparacao["Diferença"] = comparacao["Otimizado"] - comparacao["Original"]

display(comparacao.round(4))

comparacao[["Original", "Otimizado"]].plot(
    kind="bar",
    figsize=(10, 5),
    ylim=(0, 1),
    title="Árvore de Decisão: Original vs Otimizada",
    rot=0,
    grid=True
)

plt.ylabel("Pontuação")
plt.show()
```

```python
def plotConfusionMatrix(name, y_true, y_pred):
    ConfusionMatrixDisplay.from_predictions(
        y_true,
        y_pred,
        display_labels=["Sem SOP", "Com SOP"],
        cmap="Blues",
        values_format="d"
    )

    plt.title(f"Matriz de Confusão - {name}")
    plt.xlabel("Valor Predito")
    plt.ylabel("Valor Real")
    plt.show()

def evaluateModel(name, y_true, y_pred):
    print(f"\n{name}")
    print("-" * 40)

    print(
        f"Accuracy:  {accuracy_score(y_true, y_pred):.3f}"
    )

    print(
        f"Precision: {precision_score(y_true, y_pred):.3f}"
    )

    print(
        f"Recall:    {recall_score(y_true, y_pred):.3f}"
    )

    print(
        f"F1-score:  {f1_score(y_true, y_pred):.3f}"
    )

    print("\nMatriz de confusão:")
    plotConfusionMatrix(name, y_true, y_pred)
    
```

```python

import time
import pandas as pd
import matplotlib.pyplot as plt

from src.escalabilidade import Escalabilidade

# Simula uma solicitação de processamento
def processar_solicitacao():
    time.sleep(0.5)


escalabilidade = Escalabilidade(
    minimo=1,
    maximo=5,
    tarefas_por_worker=2
)

inicio = time.perf_counter()

# Simula aumento repentino de demanda
for _ in range(30):
    escalabilidade.enviar(processar_solicitacao)

escalabilidade.aguardar()
duracao = time.perf_counter() - inicio

escalabilidade.encerrar()

print(f"30 solicitações processadas em {duracao:.2f} segundos")
print(f"Máximo de workers observado: {max(h['workers'] for h in escalabilidade.historico)}")

historico = pd.DataFrame(escalabilidade.historico)
historico["segundos"] = historico["tempo"] - historico["tempo"].iloc[0]

plt.figure(figsize=(10, 4))
plt.plot(historico["segundos"], historico["workers"], label="Workers")
plt.plot(historico["segundos"], historico["pendentes"], label="Fila")
plt.xlabel("Tempo (segundos)")
plt.ylabel("Quantidade")
plt.title("Escalabilidade automática conforme a demanda")
plt.legend()
plt.grid(True)
plt.show()
```

### Resultados da Escalabilidade Automática

Para validar o mecanismo de escalabilidade, foi simulada uma carga de 30 solicitações, cada uma com duração aproximada de 0,5 segundo.

O controlador foi configurado para operar com um mínimo de 1 e um máximo de 5 workers, ajustando automaticamente a capacidade conforme a quantidade de solicitações pendentes e em execução.

**Resultados observados:**

- Solicitações processadas: 30.
- Tempo total de execução: 3,90 segundos.
- Quantidade máxima de workers: 5.
- Quantidade final de workers: 1.
- Solicitações pendentes ao final: 0.

O gráfico demonstra que, durante o aumento da demanda, o sistema expandiu automaticamente sua capacidade de processamento. Conforme as solicitações foram concluídas, a fila diminuiu e os workers excedentes foram encerrados.

Em uma execução sequencial teórica, as 30 solicitações demandariam aproximadamente 15 segundos. A execução concorrente apresentou uma redução aproximada de 74% no tempo total, considerando também o período de espera utilizado na simulação.

Os resultados demonstram o funcionamento do mecanismo de escalabilidade automática em ambiente local. A solução é adequada para ilustrar o processamento concorrente de operações com tempo de espera, mas não representa uma avaliação de desempenho com requisições reais à API do Gemini.
